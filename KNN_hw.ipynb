```python
# https://github.com/pythonai200425/pages_ai/blob/main/knn_hw.md

import numpy as np
from sklearn.neighbors import KNeighborsRegressor

# Data
X = np.array([0.1, 0.3, 0.5, 0.8, 1.1, 1.5, 1.9, 2.3, 2.8, 3.3, 3.7, 4.2]).reshape(-1, 1)
Y = np.array([1.0, 1.3, 1.5, 2.0, 2.4, 2.9, 3.3, 3.8, 4.4, 5.0, 5.4, 6.0])

# Build KNN regressor
knn = KNeighborsRegressor(n_neighbors=3)
knn.fit(X, Y)
'''
for [x_3]x = 0.5 nearest X [0.3, 0.5, 0.8] nearest Y [1.3,1.5,2.0] calc 1.3+1.5+2.0=4.8 / 3 = Y_pred = (1.6)
'''

# Predict for the same X values
Y_pred = knn.predict(X)
mse = np.mean((Y - Y_pred) ** 2)
'''
MSE = [ (Y₁ − Y_pred,₁)² + (Y₂ − Y_pred,₂)² + (Y₃ − Y_pred,₃)² ] / 3 - example
         1  - 1.2667     +  1.3 - 1.2667    +  1.5 - 1.6  ...... / n (count of Y)

Sum of squared errors = 0.3811
MSE = 0.3811 / 12 = 0.03175
'''
print("Predicted Y:", Y_pred)
print()
print("MSE:", mse)
```
